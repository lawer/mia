# Problema de las jarras de agua

Tenemos una jarra de 3 litros y otra de 4. Podemos llenarlas, vaciarlas o trasvasar agua de una a otra. El objetivo es llegar a un estado en el que una de las jarras contenga exactamente 2 litros.

## Representación del estado

Un estado es la tupla `(botella_3, botella_4)`. Por ejemplo, `(3, 1)` significa que la jarra de 3 litros está llena y la de 4 contiene 1 litro. Las tuplas son inmutables y pueden utilizarse directamente en un `set` de visitados y en el diccionario de estados anteriores.

## Estrategia de búsqueda

Todas las acciones tienen el mismo coste: un paso. Por eso utilizaremos búsqueda en anchura (BFS). BFS explora primero los estados que están a una acción de distancia, luego los que están a dos, y así sucesivamente; por tanto, encuentra una solución con el menor número de acciones.

El diccionario `pares` guarda desde qué estado se ha llegado a cada sucesor. Al final permite reconstruir la ruta sin almacenar el estado anterior dentro del propio estado.

Como en los apuntes, `frontera` es una cola `deque` y un estado pasa a `visitats` cuando se extrae. El diccionario `pares` es la única ampliación: guarda el primer predecesor de cada estado para reconstruir el camino.

```python
def succesors(estat):
    bot_3, bot_4 = estat
    estats = []

    ### PLENAR UNA BOTELLA
    if bot_3 != 3:
        estats.append((3, bot_4))

    if bot_4 != 4:
        estats.append((bot_3, 4))

    ### BUIDAR UNA BOTELLA
    if bot_3 != 0:
        estats.append((0, bot_4))

    if bot_4 != 0:
        estats.append((bot_3, 0))

    ### TRASPASSAR
    if bot_3 != 0:
        cap = 4 - bot_4
        a_passar = min(bot_3, cap)
        estats.append((bot_3 - a_passar, bot_4 + a_passar))

    if bot_4 != 0:
        cap = 3 - bot_3
        a_passar = min(bot_4, cap)
        estats.append((bot_3 + a_passar, bot_4 - a_passar))

    return (estats)
```

```python
import collections


def es_solucio(estat):
    return estat[0] == 2 or estat[1] == 2


def cerca_amplada(estat_inicial):
    frontera = collections.deque([estat_inicial])
    visitats = set()
    pares = {estat_inicial: None}

    while frontera:
        estat = frontera.popleft()
        if estat in visitats:
            continue
        visitats.add(estat)

        if es_solucio(estat):
            return estat, pares

        for succesor in succesors(estat):
            if succesor not in visitats:
                if succesor not in pares:
                    pares[succesor] = estat
                frontera.append(succesor)

    return None, pares


def reconstrueix_ruta(pares, final):
    ruta = []
    while final is not None:
        ruta.append(final)
        final = pares[final]
    return list(reversed(ruta))
```

```python
inicial = (0, 0)
final, pares = cerca_amplada(inicial)

if final is None:
    print("No se ha encontrado ninguna solución.")
else:
    for estat in reconstrueix_ruta(pares, final):
        print(estat)
```
