# Problema de les botelles d'aigua

Tenim una botella de 3 litres i una altra de 4 litres. Podem omplir, buidar o traspassar aigua entre elles. L'objectiu és arribar a un estat que continga exactament 2 litres en una de les botelles.

## Representació de l'estat

Un estat és la tupla `(botella_3, botella_4)`. Per exemple, `(3, 1)` significa que la botella de 3 litres és plena i la de 4 litres conté 1 litre. Les tuples són immutables i es poden usar directament en un `set` de visitats i en el diccionari de pares.

## Estratègia de cerca

Totes les accions tenen el mateix cost: un pas. Per això usem cerca en amplària o BFS. BFS explora primer els estats a una acció de distància, després els que estan a dues accions, i així successivament; per tant, troba una solució amb el menor nombre d'accions.

El diccionari `pares` guarda des de quin estat s'ha arribat a cada successor. Al final permet reconstruir la ruta sense guardar el pare dins de l'estat.

Com als apunts, `frontera` és una cua `deque` i un estat passa a `visitats` quan s’extrau. El diccionari `pares` és l’única ampliació: guarda el primer predecessor de cada estat per reconstruir el camí.

```python
def succesors(estat):
    bot_3, bot_4 = estat
    estats = []

    ### PLENAR UNA BOTELLA
    if bot_3 != 3:
        estats.append((3, bot_4))

    if bot_4 != 4:
        estats.append((bot_3, 4))

    ### BUIDAR UNA BOTELLA
    if bot_3 != 0:
        estats.append((0, bot_4))

    if bot_4 != 0:
        estats.append((bot_3, 0))

    ### TRASPASSAR
    if bot_3 != 0:
        cap = 4 - bot_4
        a_passar = min(bot_3, cap)
        estats.append((bot_3 - a_passar, bot_4 + a_passar))

    if bot_4 != 0:
        cap = 3 - bot_3
        a_passar = min(bot_4, cap)
        estats.append((bot_3 + a_passar, bot_4 - a_passar))

    return (estats)
```

```python
import collections


def es_solucio(estat):
    return estat[0] == 2 or estat[1] == 2


def cerca_amplada(estat_inicial):
    frontera = collections.deque([estat_inicial])
    visitats = set()
    pares = {estat_inicial: None}

    while frontera:
        estat = frontera.popleft()
        if estat in visitats:
            continue
        visitats.add(estat)

        if es_solucio(estat):
            return estat, pares

        for succesor in succesors(estat):
            if succesor not in visitats:
                if succesor not in pares:
                    pares[succesor] = estat
                frontera.append(succesor)

    return None, pares


def reconstrueix_ruta(pares, final):
    ruta = []
    while final is not None:
        ruta.append(final)
        final = pares[final]
    return list(reversed(ruta))
```

```python
inicial = (0, 0)
final, pares = cerca_amplada(inicial)

if final is None:
    print("No s'ha trobat cap solució.")
else:
    for estat in reconstrueix_ruta(pares, final):
        print(estat)
```
