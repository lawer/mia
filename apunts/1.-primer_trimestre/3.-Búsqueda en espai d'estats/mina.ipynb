# La cursa de la mina

Programa un robot autònom que arribe al tresor abans que el seu rival. Les mines del torneig es generaran després de tancar els lliuraments. El teu robot rep el mapa complet de la partida i ha de decidir què fer en cada torn.

- Els bots es mouen alhora i poden compartir casella. Si arriben al tresor en el mateix torn, empaten. Després de 60 torns sense cap arribada també hi ha empat.
- Una casella normal (`.`) consumeix 1 d'energia i el terreny difícil (`:`) en consumeix 3. Els murs (`#`) són infranquejables. Esperar no consumeix energia.
- El primer bot que agafa la bomba deixa l'altre sense moure's durant els dos torns següents. Si els dos hi arriben alhora, la bomba es gasta i no retarda ningú.

## Prova el motor

Executa les cel·les i canvia la llavor de pràctica per veure altres mines. Amb `mostra_torn(n)` pots inspeccionar qualsevol torn.

```
from pathlib import Path
import sys
from urllib.request import urlretrieve

carpeta = Path("mina")
carpeta.mkdir(exist_ok=True)
for nom in ("mina.py", "bot_model.py"):
    fitxer = carpeta / nom
    if not fitxer.exists():
        urlretrieve("https://raw.githubusercontent.com/lawer/mia/main/apunts/1.-primer_trimestre/3.-B%C3%BAsqueda%20en%20espai%20d%27estats/mina/" + nom, fitxer)
sys.path.insert(0, str(carpeta.resolve()))

from mina import dibuixa, genera_mina, juga, successors
from bot_model import decideix as bot_model
```

```
mina = genera_mina(3)
guanyador, historial = juga(mina, (bot_model, bot_model))

def mostra_torn(torn):
    robots, disponible = historial[torn]
    print(dibuixa(mina, robots, disponible))
    for i, robot in enumerate(robots, 1):
        print(f"Bot {i} | 🔋 {robot.energia} | ⏳ {robot.retard}")

mostra_torn(0)
print()
mostra_torn(len(historial) - 1)
```
