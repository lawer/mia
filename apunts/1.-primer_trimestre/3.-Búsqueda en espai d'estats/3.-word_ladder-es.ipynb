## Word Ladder

Primero implementaremos un juego sencillo llamado *Word Ladder*, con estas reglas:

Se dan dos palabras con el mismo número de letras (utilizaremos palabras de tres o cuatro letras), por ejemplo, DOG y CAT. El objetivo es transformar la primera palabra en la segunda cambiando una letra cada vez por cualquier otra, siempre que el resultado sea una palabra inglesa válida. Por ejemplo, podemos cambiar DOG por FOG, pero no por GOG (no es una palabra inglesa válida) ni por GOD (se cambian dos letras). Una ruta posible es:

**DOG => COG => COT => CAT**

También hay otras rutas, por ejemplo:

**DOG => HOG => HOT => HAT => CAT**

El objetivo del juego es encontrar la secuencia más corta de cambios que transforma la primera palabra en la segunda. Utilizaremos un diccionario de inglés para comprobar si cada palabra es válida. El archivo `words34.txt.gz` contiene todas las palabras inglesas de tres y cuatro letras, una por línea.

El juego puede generalizarse a palabras de cualquier longitud, pero en este ejercicio nos limitaremos a palabras de tres y cuatro letras.

Definiremos tres medidas de coste para el algoritmo de búsqueda:

* **Distancia**: número de cambios de letra realizados: 0 para la palabra inicial, 1 para la siguiente, etc.
  * El coste de **DOG => COG => COT => CAT** es 3.
* **Scrabble**: asignamos a cada letra el valor entero que tiene en Scrabble.
  * A, E, I, O, U, L, N, S, T y R valen 1.
  * D y G valen 2.
  * B, C, M y P valen 3.
  * F, H, V, W e Y valen 4.
  * K vale 5.
  * J y X valen 8.
  * Q y Z valen 10.
  * El coste de **DOG => COG => COT => CAT** es 3 + 1 + 1 = 5.
* **Rareza**: el tercer coste depende de lo común que sea cada palabra después de la inicial. Pasar a una palabra muy rara debería costar más que pasar a una común. Daremos una lista de las palabras inglesas más frecuentes y su **frecuencia relativa**.
  * El coste de la secuencia **DOG => JOG => JOT => COT => CAT** es 1 + R por cada paso, donde R mide la rareza de la palabra introducida. El comando Unix `zmore` permite consultar `words34.txt.gz`; por ejemplo:
    ```
    jog  10.052741
    jot  11.177437
    cot  10.071238
    cat   6.886906
    ```
    Por tanto, el coste es 4 más la suma de estos valores: 42.188322.
