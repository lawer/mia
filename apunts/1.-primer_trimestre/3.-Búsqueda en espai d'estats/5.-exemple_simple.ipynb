# Rescat en un parc nacional

Un equip vol anar del punt A al punt I pel camí de menor cost. Cada nombre representa el cost d'un trajecte; les dades són inventades. Com que els costos són diferents, utilitzarem cerca de cost uniforme (UCS).

```python
# Definició del graf amb costos d'acció
# Cada clau representa un node i el seu valor és un diccionari amb els nodes veïns i el cost per arribar-hi

graf = {
    "A": {"B": 4, "C": 3},
    "B": {"A": 4, "D": 5, "E": 12},
    "C": {"A": 3, "F": 7},
    "D": {"B": 5, "E": 4, "G": 10},
    "E": {"B": 12, "D": 4, "G": 2, "H": 5},
    "F": {"C": 7, "H": 6},
    "G": {"D": 10, "E": 2, "I": 1},
    "H": {"E": 5, "F": 6, "I": 3},
    "I": {"G": 1, "H": 3}
}
```

## Solució

La cua de prioritat tria el cost acumulat més menut. `costos` guarda el millor cost conegut i `pares` permet reconstruir el camí. Si apareix un camí més barat a un node, actualitzem tots dos.

```python
import heapq

def ucs(graf, inicial, final):
    frontera = [(0, inicial)]
    costos = {inicial: 0}
    pares = {inicial: None}

    while frontera:
        cost, node = heapq.heappop(frontera)
        if cost != costos[node]:
            continue
        if node == final:
            return cost, pares
        for vei, cost_trajecte in graf[node].items():
            nou_cost = cost + cost_trajecte
            if vei not in costos or nou_cost < costos[vei]:
                costos[vei] = nou_cost
                pares[vei] = node
                heapq.heappush(frontera, (nou_cost, vei))
    return None, pares


def reconstrueix_ruta(pares, final):
    if final not in pares:
        return []
    ruta = []
    while final is not None:
        ruta.append(final)
        final = pares[final]
    return list(reversed(ruta))
```

```python
cost, pares = ucs(graf, "A", "I")
print(" → ".join(reconstrueix_ruta(pares, "I")))
print("Cost:", cost)
```
