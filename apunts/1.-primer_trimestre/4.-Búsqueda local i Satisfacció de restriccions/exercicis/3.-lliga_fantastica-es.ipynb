# Equipo de fútbol

## Planteamiento
Tenemos que formar un equipo de fútbol con 11 jugadores. Disponemos de una lista de jugadores y, para cada uno, conocemos sus posiciones (`club_position`), su valoración (`overall`) y su precio (`value_eur`). El valor del equipo es la suma de las valoraciones de sus jugadores. Nuestro objetivo es formar el equipo de mayor valor posible, respetando estas restricciones:

- Obligatoria:
    - El precio del equipo no puede superar los 250 millones de euros.
- Restricciones adicionales:
    - Solo podemos tener 4 defensas, 4 centrocampistas y 2 delanteros.
    - El valor total del equipo no puede superar los 250 millones de euros.
    - Solo podemos tener 3 jugadores del mismo equipo.
    - Solo podemos tener 2 jugadores de la misma liga.
    - Solo podemos tener 3 jugadores de la misma nacionalidad.
    - Solo podemos tener 1 portero.
    - Solo podemos tener 1 jugador de la misma posición.

Utiliza un algoritmo genético para encontrar el mejor equipo posible.

1.- Carga los datos y muestra las 5 primeras filas.

2.- Muestra el número de filas y columnas del conjunto de datos.

3.- Muestra las distintas posiciones de los jugadores e indica cuántos hay en cada posición.

4.- Elimina los jugadores que no tengan valoración.

5.- Utilizando un algoritmo genético, encuentra el mejor equipo posible. El precio del equipo no puede superar los 250 millones de euros.
