- Cada equipo jugará 18 partidos regulares.
- Cada equipo jugará 9 partidos como local y 9 como visitante.
- Cada equipo jugará dos partidos contra cada equipo de su grupo.

- Los equipos alternarán una semana como locales y otra como visitantes.
