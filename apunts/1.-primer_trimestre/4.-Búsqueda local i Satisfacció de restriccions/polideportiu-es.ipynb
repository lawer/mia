# Planificación de horarios de un polideportivo

Cada equipo puede recibir como máximo una reserva. Dos reservas no pueden usar la misma pista en la misma franja. Buscamos un horario que sume tantos puntos de preferencia como sea posible.

Por ejemplo, si un equipo reserva una pista de 10 a 12, ningún otro puede reservar esa pista durante la misma franja. Cada preferencia puntúa de 1 a 5.

## Descripción del problema

Empezaremos definiendo los equipos que quieren reservar las pistas. Cada equipo tiene un identificador y un nombre.

```python
noms_equips = [
    "Els Parats del Onze",
    "Els Xiquets del Barri",
    "Els Súper Campions del Parc",
    "Els Futbolistes de Cap de Setmana",
    "L'Onze Ideal del Bar",
    "Els Cracks de la Pista",
    "Els Invencibles del Carrer",
    "Els Reis del Partidazo",
    "Els Mags de la Pilota",
    "Els Golejadors Nats",
    "Els Artistes del Futbol",
    "Els Malasangre FC",
    "Els Veterans Imparables",
    "Els Joves Promeses",
    "Els Amics de la Pilota",
    "Els Galàctics del Barri",
    "Els Pirates del Futbol",
    "Els Ninjes de la Pilota",
    "Els Zombies del Futbol",
    "Els Superherois de l'Esport",
    "Els Capos del Barri",
    "Els Pames de la Pista",
    "Els Toreros de la Pilota",
    "Els Vikings del Futbol",
    "Els Samurais de la Pilota",
    "Els Cavallers de la Redona",
    "Els Guerrers Espartans",
    "Els Gladiadors de la Pista",
    "Els Lleons del Barri",
    "Els Tigres del Futbol",
    "Els Dracs de la Pilota",
    "Les Àguiles del Barri",
    "Els Hòldings del Futbol",
    "Els Voltors de la Pilota",
    "Els Llops del Barri",
    "Els Raors del Futbol",
    "Els Ossos de la Pilota",
    "Els Conills de la Pista",
    "Els Ratolins del Barri",
    "Els Gats del Futbol",
    "Els Cans de la Pilota",
    "Els Porcs de la Pista",
    "Els Polls del Barri",
    "Els Galls del Futbol",
    "Els Pavos Reals de la Pilota",
    "Els Ànecs de la Pista",
]

for i in range(0, len(noms_equips)):
    print(f"Equip {i+1}: {noms_equips[i]}")
```

A continuación, definiremos las pistas del polideportivo. Cada una tiene un identificador, un nombre y una capacidad.

```python
pistes = [
    {"nom": "Pista 1", "capacitat": 100},
    {"nom": "Pista 2", "capacitat": 75},
    {"nom": "Pista 3", "capacitat": 150},
    {"nom": "Pista 4", "capacitat": 200},
]
```

Por último, definiremos las preferencias de cada equipo: las pistas que quiere reservar y el intervalo horario que prefiere. El polideportivo solo abre los sábados y domingos de 8:00 a 22:00, y los viernes de 16:00 a 22:00.

```python
from random import Random

rng = Random(42)

# Definim els horaris de les pistes

horaris = [
    "divendres de 16:00 a 18:00", "divendres de 18:00 a 20:00", "divendres de 20:00 a 22:00", "dissabte de 08:00 a 10:00", "dissabte de 10:00 a 12:00", "dissabte de 12:00 a 14:00", "dissabte de 16:00 a 18:00", "dissabte de 18:00 a 20:00", "dissabte de 20:00 a 22:00", "diumenge de 08:00 a 10:00", "diumenge de 10:00 a 12:00", "diumenge de 12:00 a 14:00", "diumenge de 16:00 a 18:00", "diumenge de 18:00 a 20:00", "diumenge de 20:00 a 22:00"
]

# Creem les preferències de cada equip. Cada equip tindrà una llista de tuples amb les seves preferències. Cada tupla tindrà el nom de la pista, l'horari i la preferència (de 1 a 5)

preferencies = []

for i in range(0, len(noms_equips)):
    equip = noms_equips[i]
    preferencies_equip = []
    for j in range(0, 5):
        pista = rng.choice(pistes)
        horari = rng.choice(horaris)
        preferencia = rng.randint(1, 5)
        preferencies_equip.append((pista["nom"], horari, preferencia))
    preferencies.append((equip, preferencies_equip))

print(preferencies[0])
```

## Del orden de los equipos a un horario
El estado es una tupla con los índices de los equipos en el orden en que los atenderemos. Cada equipo elige su franja libre mejor puntuada. Si todas sus opciones están ocupadas, se queda sin reserva. Así, cualquier orden genera un horario factible.

```python
def planifica(ordre):
    ocupades = set()
    reserves = []
    punts = 0
    for i in ordre:
        equip, opcions = preferencies[i]
        for pista, horari, valor in sorted(opcions, key=lambda opcio: -opcio[2]):
            if (pista, horari) not in ocupades:
                ocupades.add((pista, horari))
                reserves.append((equip, pista, horari))
                punts += valor
                break
    return reserves, punts

inicial = tuple(range(len(preferencies)))
reserves_inicials, punts_inicials = planifica(inicial)
print(punts_inicials, reserves_inicials[:3])
```
