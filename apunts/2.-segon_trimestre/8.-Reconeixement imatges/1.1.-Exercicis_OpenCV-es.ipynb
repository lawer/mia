# Ejercicios de OpenCV

Hemos visto algunas funciones básicas de OpenCV. Ahora las pondremos en práctica.

Aplicarás lo aprendido sobre procesamiento de imágenes en una actividad parecida a la que hemos hecho en clase.

## Ejercicio 2
