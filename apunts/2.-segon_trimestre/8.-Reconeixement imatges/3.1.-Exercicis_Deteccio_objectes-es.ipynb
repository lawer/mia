# Ejercicios de detección de objetos

En esta práctica aplicarás las técnicas del ejemplo de YOLO para detectar objetos en imágenes.

Busca un conjunto de datos que te interese y utiliza la detección de objetos para resolver un problema concreto.

Si no tienes una idea, puedes probar alguna de estas propuestas:

- [Detectar señales de tráfico para ayudar a la conducción autónoma](https://universe.roboflow.com/road-guard/road-sign-detector-jpryp).
- [Detectar cartas para identificar las manos de una partida de póquer](https://universe.roboflow.com/roboflow-100/poker-cards-cxcvz).
- [Detectar enemigos en un videojuego para programar un bot que juegue automáticamente](https://universe.roboflow.com/graham-doerksen/halo-infinite-angel-aim).

Estos ejemplos ofrecen un modelo funcional que puedes utilizar para comparar los resultados con los de tu propio modelo.
