# Clasificación de imágenes

Hemos visto cómo entrenar una red neuronal para clasificar imágenes.

Ahora trabajaremos con otro conjunto de datos para comprobar si lo aprendido se aplica también a otros datos.

Utilizaremos [HES-XPLAIN/SportsImageClassification](https://huggingface.co/datasets/HES-XPLAIN/SportsImageClassification), de Hugging Face. Contiene imágenes de distintos deportes. El objetivo es clasificarlas y evaluar cómo se comporta la red con imágenes que no ha visto durante el entrenamiento.

Puedes utilizar cualquiera de las técnicas vistas hasta ahora (redes *feed-forward*, redes convolucionales, *transfer learning* o modelos preentrenados) para obtener el mejor resultado posible. Intenta conseguir una *validation accuracy* superior al 80 %.

Si no lo consigues, prueba la técnica de *data augmentation* y comprueba si mejora los resultados.
