# Sistema de procesamiento de texto para una tarea específica

En esta práctica implementarás un sistema de procesamiento de texto para una tarea concreta: controlar un robot mediante la voz.

Utiliza las herramientas de procesamiento de texto que has aprendido para crear un sistema que reciba una orden de movimiento y devuelva la instrucción que debe ejecutar el robot.

El sistema debe cumplir estas especificaciones:

- El robot debe poder **moverse** a una velocidad de 1 a 10.
- El robot debe poder **girar** a la izquierda o a la derecha.
- El robot debe poder **pararse**; en ese caso, no realizará ningún movimiento.
- Debe ser posible especificar **distintos verbos** para una misma acción (por ejemplo, «avanza» en lugar de «mueve»).
- Debe ser posible indicar más de una orden en una frase (por ejemplo, «avanza 5 y gira a la izquierda»).

Sigue estos pasos para implementar el sistema:

1. **Recibir los datos de entrada:** recibirás una cadena de texto con las órdenes para el robot; puede contener más de una.
2. **Eliminar las palabras innecesarias:** utiliza herramientas de procesamiento de texto para eliminar las palabras que no hacen falta (*stopwords*).
3. **Tokenizar y etiquetar las palabras:** divide el texto en tokens y asígnales etiquetas.
4. **Identificar los verbos:** localiza los verbos que expresan órdenes para el robot.
5. **Identificar las direcciones:** determina hacia dónde debe moverse el robot.
6. **Identificar los números:** localiza los números que indican cuánto debe moverse.
7. **Generar la instrucción:** construye la orden que debe ejecutar el robot.
