# Clasificar preguntas

En la práctica 2 vimos cómo convertir un texto en una representación numérica que pueda utilizar un algoritmo de aprendizaje automático. Probamos distintas representaciones, como *Bag of Words* y *word embeddings*, y entrenamos una red neuronal para clasificar textos.

En esta práctica repetirás el proceso para clasificar preguntas por temas. Utilizaremos el conjunto de datos `trec`, que contiene preguntas en inglés con su temática. Está disponible en [Hugging Face Datasets](https://huggingface.co/datasets/CogComp/trec).

## Objetivos de la práctica

* Reproducir el proceso de la práctica 2 para clasificar preguntas por temas.
* Preparar una red neuronal que clasifique las preguntas.
* Probar distintas representaciones numéricas del texto.
* Comparar los resultados obtenidos con cada representación.
* Comparar también los resultados con los de un modelo basado en redes neuronales recurrentes.
