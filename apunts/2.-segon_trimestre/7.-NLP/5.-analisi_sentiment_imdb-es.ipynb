# Análisis de sentimientos con DistilBERT

En la práctica 2 entrenamos DistilBERT para clasificar sentimientos en textos a partir de un conjunto de tuits.

En esta práctica aplicaremos el modelo a reseñas de películas. El entrenamiento anterior no sirve directamente para esta nueva tarea, así que utilizaremos *transfer learning* con el conjunto de reseñas de IMDB, disponible en Hugging Face con el nombre `imdb`.

## Objetivos de la práctica

* Reproducir el proceso de la práctica 3 para entrenar DistilBERT —u otro modelo Transformer— en la clasificación de sentimientos de reseñas de películas.
* Analizar las predicciones y compararlas con las del modelo entrenado en la práctica 2.
* Utilizar un `pipeline` de Transformers para clasificar reseñas mediante [`zero-shot`](https://huggingface.co/docs/transformers/main_classes/pipelines#transformers.ZeroShotClassificationPipeline.example) y comparar los resultados con los del modelo entrenado con IMDB.
