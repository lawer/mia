# Lesiones de rodilla

Las lesiones de rodilla son frecuentes en deportes como el fútbol, el baloncesto o el esquí. Pueden ser de varios tipos; entre las más comunes están las lesiones del ligamento cruzado anterior (LCA) y del menisco.

Sus síntomas a menudo se parecen, pero el tratamiento es distinto. Por eso es importante identificar la lesión para aplicar el tratamiento adecuado y evitar que empeore.

Crearemos un sistema experto con `experta` para ayudar a entrenadores y personal médico no especializado a orientar el diagnóstico de lesiones de rodilla, a partir de casos diagnosticados por especialistas.

El sistema se basa en el trabajo [An Expert System for Knee Problems Diagnosis](https://www.researchgate.net/publication/354150948_An_Expert_System_for_Knee_Problems_Diagnosis/link/6127bc8338818c2eaf62917b/download?_tp=eyJjb250ZXh0Ijp7ImZpcnN0UGFnZSI6InB1YmxpY2F0aW9uIiwicGFnZSI6InB1YmxpY2F0aW9uIn19) de [Lamis F. Samhan](https://www.researchgate.net/profile/Lamis-Samhan) y [Samy S. Abu-Naser](https://www.researchgate.net/profile/Samy-Abu-Naser).

## Hechos

Para inferir una posible lesión, el sistema experto tiene en cuenta los siguientes síntomas:

- `colp` (s1): el dolor o la inflamación de la rodilla comenzó después de una caída, una torsión o un golpe recibido de un objeto o de otra persona.
- `deformacio` (s2): la rodilla está deformada.
- `rotula inflamada` (s3): la rótula está inflamada, sensible y caliente, y duele al realizar actividad.
- `genoll inflamat` (s4): la rodilla está inflamada, sensible y caliente; duele después de estar sentado un rato o de subir y bajar escaleras.
- `dolor posterior` (s5): hay un dolor agudo en la parte posterior de la rodilla y duele al estirar la pierna.
- `crepitacions` (s6): se oyen crujidos en la rodilla o esta se bloquea, aunque el dolor mejore.
- `roig` (s7): la rodilla está enrojecida o inflamada.
- `febra` (s8): hay fiebre junto con la rodilla inflamada o enrojecida.
- `temps` (s9): la rodilla se ha debilitado con el tiempo y el dolor empeora cuando cambia el tiempo.
- `inflacio_darrera` (s10): hay inflamación o sensibilidad en la parte posterior de la rodilla.
- `jove_baix_rotula` (s11): el paciente tiene entre 12 y 18 años y presenta dolor en la parte anterior de la rodilla que empeora con la actividad.
- `jove_genoll` (s12): el paciente tiene menos de 18 años y presenta dolor de rodilla.

Según la combinación de síntomas, el sistema puede sugerir las siguientes posibilidades:

- Si el paciente tiene `colp`:
  - junto con `deformacio`, puede sugerir una `fractura` o un `lligament trencat`;
  - junto con `rotula inflamada`, puede sugerir una `fractura rotula`;
  - junto con `genoll inflamat`, puede sugerir un `lligament trencat` o un `menisc trencat`;
  - junto con `dolor posterior`, puede sugerir un `múscul isquiotibial trencat`;
  - junto con `crepitacions`, puede sugerir un `menisc trencat`.
- Si tiene `roig` y `febra`, puede sugerir `artritis reumatoide`.
- Si tiene `temps`, puede sugerir `osteo artritis`.
- Si tiene `inflacio_darrera`, puede sugerir `quist de Baker`.
- Si tiene `jove_baix_rotula`, puede sugerir `sindrome de Osgood-Schlatter`.
- Si tiene `jove_genoll`, puede sugerir `dolor de creixement`.
